Quartis

```python
def calcula_quartis(lista):
    lista_ordenada = sorted(lista)
    tamanho = len(lista)

    posicao_q1 = 1 * (tamanho + 1) // 4 
    posicao_q2 = 2 * (tamanho + 1) // 4 
    posicao_q3 = 3 * (tamanho + 1) // 4 

    q1 = lista_ordenada[int(posicao_q1) -1]
    q2 = lista_ordenada[int(posicao_q2) -1]
    q3 = lista_ordenada[int(posicao_q3) -1]

    return q1, q2, q3

vendas = [42,18,35,27,50,31,22]

calcula_quartis(vendas)
```

```python
def calcular_quartis(valores):
    dados_ordenados = sorted(valores) # valores ordenados
    n = len(dados_ordenados) # tamanho do conjunto de dados
    
    posicoes = [
        (n + 1) / 4,
        (n + 1) / 2,
        (3 * (n + 1)) / 4
    ]
    
    quartis = []
    
    for posicao in posicoes:
        posicao_inteira = int(posicao)
        
        if posicao == posicao_inteira:
            valor_quartil = dados_ordenados[posicao_inteira - 1]
            
        else:
            parte_decimal = posicao - posicao_inteira
            
            valor_inferior = dados_ordenados[posicao_inteira - 1]
            valor_superior = dados_ordenados[posicao_inteira]
            
            diferenca = valor_superior - valor_inferior
            proporcao = diferenca * parte_decimal
            
            valor_quartil = valor_inferior + proporcao
            
        quartis.append(valor_quartil)
        
    return quartis

vendas = [42, 18, 35, 30, 27, 50, 31, 22]

q1, q2, q3 = calcular_quartis(vendas)

print(f"Q1: {q1}")
print(f"Q2: {q2}")
print(f"Q3: {q3}")
```

### Utilizando a biblioteca Statistics

### Média

```python
eqp_a= 42,38,45,41,39,44,38
eqp_b= 30,52,35,48,29,55,38

media_a=statistics.mean(eqp_a)
media_b=statistics.mean(eqp_b)
print(f'a média da equipe A é de: {media_a}')
print(f'a média da equibe B é de: {media_b}')
```

### Mediana

```python
mediana_a=statistics.median(eqp_a)
mediana_b=statistics.median(eqp_b)
print(f'a média da equipe A é de: {mediana_a}')
print(f'a média da equibe B é de: {mediana_b}')
```

### Moda

```python
moda_a=statistics.mode(eqp_a)
moda_b=statistics.mode(eqp_b)
print(f'a moda da equipe A é de: {moda_a}')
print(f'a moda da equibe B é de: {moda_b}')
```

### Multimodal

```python
multiModal_a=statistics.multimode(eqp_a)
multiModal_b=statistics.multimode(eqp_b)
print(f'a  da equipe A é de: {multiModal_a}')
print(f'a  da equibe B é de: {multiModal_b}')
```

### Amplitude

```python
amplitude_a= max(eqp_a) - min(eqp_a)
amplitude_b= max(eqp_b) - min(eqp_b)
print(f"a amplitude da equipe A é de: {amplitude_a}")
print(f"a amplitude da equipe B é de: {amplitude_b}")
```
