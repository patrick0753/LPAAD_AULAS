### Medida de dispersão
- Amplitude: Diferença entre o maior e o menor valor
- Variância: Mede o quanto os valores se afastam da média
- Desvio: Raíz quadrada da variância - Afastamento típico dos valores em relação à média
- Quartis: Dividem os dados ordenados em quatro partes (ex: 100 é dividido em 4 quartís de 25)
