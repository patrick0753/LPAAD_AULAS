```python
chamados = [  
    {
        "codigo": "CH001",
        "responsavel": "Maria",
        "abertura": "15/09/2026",
        "fechamento": "17/09/2026",
        "prioridade": "Alta"
    },
    {
        "codigo": "CH002",
        "responsavel": "Carlos",
        "abertura": "16/09/2026",
        "fechamento": "22/09/2026",
        "prioridade": "Média"
    },
    {
        "codigo": "CH003",
        "responsavel": "Marina",
        "abertura": "18/09/2026",
        "fechamento": "19/09/2026",
        "prioridade": "Alta"
    },
    {
        "codigo": "CH004",
        "responsavel": "Rafael",
        "abertura": "20/09/2026",
        "fechamento": "25/09/2026",
        "prioridade": "Baixa"
    },
    {
        "codigo": "CH005",
        "responsavel": "Maria",
        "abertura": "21/09/2026",
        "fechamento": "23/09/2026",
        "prioridade": "Média"
    }

]
```

### Percorrendo lista 

```python
for numero, chamado in enumerate(chamados, start=1):
    print(f"Chamado {numero}:")
    print(f"Código: {chamado['codigo']}")
    print(f"Responsável: {chamado['responsavel']}")
    print(f"Abertura: {chamado['abertura']}")
    print(f"Fechamento: {chamado['fechamento']}")
    print(f"Prioridade: {chamado['prioridade']}")
    print()
```

### Convertendo para date

```python
#convertendo lista p datetime
import datetime

for chamado in chamados:
    chamado['abertura'] = datetime.datetime.strptime(chamado['abertura'], "%d/%m/%Y")
    chamado['fechamento'] = datetime.datetime.strptime(chamado['fechamento'], "%d/%m/%Y")
```

```python
print(type(chamados[0]['abertura']))
print(type(chamados[0]['fechamento']))
```

### Calculando tempo do atendimento

```python
### calculando tempo de atendimento
for chamado in chamados:
    tempo_atendimento = chamado['fechamento'] - chamado['abertura']
    print(f"Código: {chamado['codigo']}, Tempo de Atendimento: {tempo_atendimento.days} dias")
```

### Resumão do chamado

```python
### resuminho geral do atendimento
for chamado in chamados:
    print(f"código do chamado: {chamado['codigo']}, responsável: {chamado['responsavel']}, abertura: {chamado['abertura'].strftime('%d/%m/%Y')}, fechamento: {chamado['fechamento'].strftime('%d/%m/%Y')}, prioridade: {chamado['prioridade']}, tempo de atendimento: {(chamado['fechamento'] - chamado['abertura']).days} dias")
```

### Se está ou não dentro do prazo
- **Dentro do prazo**: <=3 dias;
- **Fora do prazo**: >3 dias;

```python
### dentro ou fora do prazo? 3 dias
for chamado in chamados:
    tempo_atendimento = (chamado['fechamento'] - chamado['abertura']).days
    if tempo_atendimento <=3:
        print(f'codigo: {chamado['codigo']} dentro do prazo')
    else:
        print(f'codigo: {chamado['codigo']} fora do prazo')
```

## PT2

```python
#qtd total de atendimento:
total_atendimentos = len(chamados)
print(f"O total de atendimentos é de: {total_atendimentos}")

#qtd de atendimentos dentro e fr do prazo:
total_dentro_prazo = 0
total_fora_prazo = 0
for chamado in chamados:
    if (chamado['fechamento'] - chamado['abertura']).days <= 3:
        total_dentro_prazo += 1
    if (chamado['fechamento'] - chamado['abertura']).days > 3:
        total_fora_prazo += 1
print(f"O total de atendimentos dentro do prazo é de: {total_dentro_prazo}\nO total de atendimentos fora do prazo é de: {total_fora_prazo}")
```

```python
dentro_prazo = []
fora_prazo = []
maior_tempo = 0
cod_maior_tempo = ""
alta=[]
media=[]
baixa=[]
for chamado in chamados:
    tempo_atendimento = (chamado['fechamento'] - chamado['abertura']).days
    if tempo_atendimento <= 3:
        dentro_prazo.append(chamado['codigo'])
    else:
        fora_prazo.append(chamado['codigo'])
    if tempo_atendimento > maior_tempo:
        maior_tempo = tempo_atendimento
        cod_maior_tempo = chamado['codigo']
    if chamado['prioridade'] == 'Alta':
        alta.append(chamado['codigo'])
    elif chamado['prioridade'] == 'Média':
        media.append(chamado['codigo'])
    elif chamado['prioridade'] == 'Baixa':
        baixa.append(chamado['codigo'])
print(f"O código do chamado que apresentou o maior tempo de atendimento é: {cod_maior_tempo} com {maior_tempo} dias")
print(f'Os códigos dos chamados dentro do prazo são: {dentro_prazo}')
print(f'Os códigos dos chamados fora do prazo são: {fora_prazo}')
print(f'Os códigos dos chamados com prioridade alta são: {alta}')
print(f'Os códigos dos chamados com prioridade média são: {media}')
print(f'Os códigos dos chamados com prioridade baixa são: {baixa}')
```
