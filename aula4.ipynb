```python
def texto_colorido(texto, cor, estilo):

    #bloco de definir a cor
    if cor=="vermelho":
        codigo=31
    elif cor=="verde":
        codigo=32
    elif cor=="ciano":
        codigo=36
    else:
        codigo=0
    if estilo=="negrito":
        formatacao=1
    elif estilo=="italico":
        formatacao=3
    else:
        formatacao=0

    return f"\033[{formatacao};{codigo}m{texto}\033[0m"
```

```python
#print(f" O texto formatado é: {texto_colorido('teste texto','verde','italico')}")
print(texto_colorido("texto","vermelho","italico"))
```

```python
def carregando_sistema():
    print(texto_colorido("conectando ao sistema...","ciano","italico"))
    time.sleep(1)
    print(texto_colorido("sistema em carregamento...","vermelho","negrito"))
    time.sleep(2)
    print(texto_colorido("sistema carregado!","verde"))
    time.sleep(0)
    return carregando_sistema()
```

### Mediana

```python
def calcula_mediana(lista):
    lista_ordenada=sorted(lista)
    lista_elementos = len(lista)
    meio = lista_elementos//2

    if lista_elementos%2==0:
        mediana=(lista_ordenada[meio-1]+lista_ordenada[meio])/2
    else:
        mediana = lista_ordenada[meio]
    return mediana
```

```python
print(calcula_mediana([23,14,31,17,20]))
```

### Moda

```python
def calcula_moda(lista):
    moda=[0]
    maior_frequencia = 0
    for valor in lista:
        frequencia = lista.count(valor)
        if frequencia>maior_frequencia:
            maior_frequencia=frequencia
            moda=valor
        return moda
```

```python
print(calcula_moda([25,14,14,31,17,20,15]))
```

### +1 Moda

```python
def calcula_moda_novo(lista):
    moda=[]
    maior_frequencia = 0
    for valor in lista:
        frequencia = lista.count(valor)
        if frequencia>maior_frequencia:
            maior_frequencia=frequencia
            modas=[valor]
        elif frequencia == maior_frequencia and valor not in modas:
            modas.append(valor)
    if maior_frequencia ==1:
        return 'amodal',[]
    elif len(modas)==1:
        return 'unimodal',modas
    elif len(modas)==2:
        return "bimodal",modas
    else:
        return "multimodal",modas
```

```python
lista_1 = [10,20,20,30,40]
lista_2 = [10,10,20,20,30]
lista_3 = [10,10,20,20,30,30,40]
lista_4 = [10,20,30,40]

print(calcula_moda_novo(lista_1))
print(calcula_moda_novo(lista_2))
print(calcula_moda_novo(lista_3))
print(calcula_moda_novo(lista_4))
```
