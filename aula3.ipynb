### while + append + lista

```python
producoes = []
qtd_produzida = int(input("digite um valor entre 300 e 800 (Ou digite -1 para parar)"))

while qtd_produzida != -1:
    if qtd_produzida <300 or qtd_produzida >800:
        print("digite um valor válido: ")
        qtd_produzida = int(input("digite um valor entre 300 e 800 (Ou digite -1 para parar)"))
    else:
        producoes.append(qtd_produzida)
        qtd_produzida = int(input("digite um valor entre 300 e 800 (ou digite -1 para parar)"))

soma_producoes = sum(producoes)

for producao in range(len(producoes)):
    if producoes[producao] >= 500:
        print(f'meta atingida, o valor de produções foi de: {producoes[producao]} peças')
        soma_meta_atingida +=1
    elif producoes[producao]>400 and producoes[producao] <500:
        print(f'abaixo da meta, o valor de producões foram de: {soma_producoes}')
    else:
        print(f'critico, o valor de produções foram de: {soma_producoes}')

qtd_turno = len(producoes)
media_producoes = soma_producoes/qtd_turno
percentual_turnos = (soma_meta_atingida/qtd_turnos) *100

print(f'a quantidade de turnos foram de: {qtd_turno}')
print(media_producoes)
print(percentual_turnos)
```

### Controle de estoque

```python
disponiveis =[]
critico = []
esgotado = []
estoque = [18,5,0,32,4,7,0,14,3]
for qtd in estoque:
    if qtd >5:
        disponiveis.append(qtd)
    elif qtd <=5 and qtd >=1:
        critico.append(qtd)
    else:
        esgotado.append(qtd)
porcentagem = ((len(esgotado)+len(critico))/len(estoque)) *100 #corrigir
print(f'o percentual é de: {porcentagem:.2f}%')
print(f' quantidade de produtos esgotados: {len(esgotado)}, quantidade de produtos em nível crítico: {len(critico)}, quantidade de estoque disponível: {len(disponiveis)}')
```
