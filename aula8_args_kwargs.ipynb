## ARGS

def função (*args):

```python
def registrar_evento(origem,*recursos):
    print(f'origem: {origem}')
    for recurso in recursos:
        print(f'recurso: {recurso}')

registrar_evento(
    'servidor', 'CPU', 'Memória', 'Disco'
)
```

### Registro de participantes

```python
#utilizando enumerate
def registrar_treinamento(tema,*participantes):
    print(f'tema do treinamento: {tema}')
    print(f'quantidade de participantes {len(participantes)}')

    for i , participante in enumerate(participantes,start=1):
        print(f'{i} - {participante}')
```

```python
registrar_treinamento('Caixa de supermercado', 'memphis', 'ganso', 'neymar')
```

### Registro de Ocorrências

```python
def registrar_ocorrencia(codigo,*sistemas,**informacoes):
    print(f'código da ocorrência: {codigo}')
    print(f'qtd de sistemas afetados {len(sistemas)}')
    for i, sistema in enumerate(sistemas,start=1):
        print(f'Sistema {i}: {sistema}')

    print('\ninformações adicionais')

    for chave, valor in informacoes.items():
        print(f'{chave}: {valor}')
```

```python
sistemas = [
    "ServidorWeb",
    "BancoDados",
    "API"]
dados = {
    "prioridade": "alta",
    "responsavel": "Ana",
    "status": "em análise"}
registrar_ocorrencia(
    "INC-001",
    *sistemas,
    **dados)
```

### Problema 1: Análise de vendas por vendedor

```python
import statistics
def registro_vendas(nome,*vendas):
    print(f'Vendedor: {nome}')
    valor_total = sum(vendas)
    media_vendas = statistics.mean(vendas)
    maxima = max(vendas)

    print(f'O Valor total de vendas foi de: {valor_total}, a média foi de {media_vendas} e o valor máximo é de {maxima}')

    if valor_total>=5000:
        print("meta atingida")
    else:
        print("meta ñ atingida") 
```

```python
registro_vendas('José',5000,4000,3000,1000,1000,2000)
registro_vendas('Ancelotti',5500,35656,45454,2000)
registro_vendas('Rodrigo',2500,3656,4454,1000)
```

### Problema 2: Registro de atendimento

```python
def registro_atendimento(pacientes,*procedimentos,**informacoes):
    print(f"paciente: {pacientes}")
    print(f'a quantidade de procedimentos foi de: {len(procedimentos)}')
    for i, procedimento in enumerate(procedimentos,start=1):
        print(f'procedimento {i} - {procedimento}')

    print('\n informações adicionais:')

    for chave, valor in informacoes.items():
        print(f'{chave}: {valor}')

    if "retorno" in informacoes:
        print(f'a data do retorno é: {informacoes['retorno']}')
    else:
        print('retorno não informado')
```

```python
registro_atendimento(
    "Neymar",
    "Cirurgia",
    "Unha encravada",
    "Migué (tem jogo sábado)",
    profissional="James Rodriguez",
    unidade="UBS Cubatão",
    convenio="SUSpeito",
    retorno= "19/05/2028"
    )
```
