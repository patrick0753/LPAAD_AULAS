### String to date - strptime() string parse time

```python
data = datetime.strptime(data_texto,"%d/%m/%Y")
print(data)
print(type(data))
```

### Exercícios
- **Conversão de string p data**

```python
data_pedido = "12/09/2026"
data_entrega = "25/12/2026"
```

```python
data_pedido_padrao = datetime.strptime(data_pedido,"%d/%m/%Y")
data_entrega_padrao = datetime.strptime(data_entrega,"%d/%m/%Y")
diferenca_dias = data_entrega_padrao-data_pedido_padrao
diferenca = diferenca_dias.days

if diferenca <5:
    print("Produto entregue dentro da data")
else:
    print(f"Produto atrasou um total de {diferenca_dias} dias")

print(f'a diferença de dias entre o pedido e a entrega é de: {diferenca_dias}')
print(f'Data do pedido: {data_pedido_padrao} || Data da entrega: {data_entrega_padrao}')
```

### Enumerate

```python
produtos = ["notebook", "mouse", "teclado"]

for numero, produto in enumerate(produtos, start=1):
    print(numero, produto)
```

```python
alunos = [
    {"nome": "felipe", "nota": 6.5},
    {"nome": "bethina", "nota": 8.0},
    {"nome": "josé", "nota": 9.2}
]

for numero, aluno in enumerate(alunos, start=1):
    print(f"{numero} - {aluno['nome']}: {aluno['nota']}")
```

### Exercicio enumerate

```python
vendedores = [
    {"nome": "ana", "vendas":12500},
    {"nome": "carlos", "vendas":9000},
    {"nome": "marina", "vendas": 15700},
    {"nome": "rafael", "vendas": 11500}
]
```

- percorrendo lista utilizando enumerate()
- numerando os vendedores
- apresentando o nome e valor de vendas de cada vendedor
- classifique cada vendedor como: meta atingida: 10k

```python
for numero, vendedor in enumerate(vendedores, start=1):
    print(f"{numero} - {vendedor['nome']}")
    if vendedor['vendas']>10000:
        print("meta atingida")
    else:
        print("meta não atingida")
```
