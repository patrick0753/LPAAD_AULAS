### Utilizando While

- enquanto verdadeiro, digite a sua idade: Se verdadeiro, pare.
- Se falso, repita. O while só para quando for verdadeiro

```python
while True:
    try:
        idade= int(input("digite sua idade: "))
        break
    except ValueError:
        print("Digite um número inteiro e positivo: ")
```

# Bloco else
- Só sai se o try der certo (espécie de break)

```python
try:
    numero = int(input("numero: "))
except ValueError:
    print("inválido")
else:
    print(f" o número digitado é o: {numero}")
```

## Finally
- Finally finaliza o processo mesmo que haja erros ou não

```python
try:
    numero = int(input("numero: "))
except ValueError:
    print("inválido")
finally:
    print("operação finalizada")
```

## Os 4 juntos

```python
try:
    # gera exceção
    numero= int(input("n1: "))
except ValueError:
    # tratamento
    print("digite um número inteiro: ")
else:
    # executado se o try for verdadeiro
    print(f'o numero digitado foi {numero}')
    # sempre executado
finally:
    print("Operação finalizada")
```

# validação com args

```python
def registrar_notas(aluno, *notas):
    if not isinstance(aluno,str) or not aluno:
        print("aluno inexistente")
        return
    if not notas:
        print("informe pelo menos uma nota: ")
        return
    for nota in notas:
        if not isinstance(nota,(int,float)):
            print("notas devem ser numéricas")
            return
        if nota <0:
            print("nota inválida")
            return

    print(f"aluno: {aluno}")
    print(f'media de notas: {sum(notas)/len(notas)}')
```

```python
registrar_notas("patrick",5,5,5,5)
```
